```python
from pymongo import MongoClient

client = MongoClient('mongodb://localhost:27017/')
db = client['company_db']
collection = db['employees']

print('Connected to MongoDB')
```

```python
# Operasi CRUD Dasar

employee_data = {
    'name': 'Alice',
    'department': 'Finance',
    'age': 29,
    'salary': 4500
}

collection.insert_one(employee_data)
print('Data inserted')

# Tugas: Insert beberapa data tambahan, lakukan update, serta delete data

employee_tambahan = [{
    'name': 'Sucipto',
    'department': 'Sales',
    'age': 37,
    'salary': 3400},{
    'name': 'Suci',
    'department': 'Finance',
    'age': 25,
    'salary': 4000},{
    'name':'Mayvandra',
    'department': 'Marketing',
    'age':27,
    'salary': 4200}]

collection.insert_many(employee_tambahan)
print('New Data inserted')


collection.update_one(
    {'name': 'Sucipto'}, {
        '$set': {'department': 'Marketing'}}
)
print('Data Sucipto was updated')


collection.delete_one({'name':'Sucipto'})
print("Data Sucipto was delete")


for employee in collection.find({}, {'_id': 0}):
    print(employee)
```

```python
pipeline = [
    {'$group': {'_id': '$department', 'average_salary': {'$avg': '$salary'}}}
]
for result in collection.aggregate(pipeline):
    print(result)
```

```python
average_price = [{'$group' : {'_id' : None, 'average': { '$avg': '$price'}}}]

result_avg = list(products.aggregate(average_price))
avg_price = result_avg[0]['average']

print(f"\nHarga rata-rata produk adalah {avg_price:.2f}")

print("Produk di atas harga rata rata")
pricey = products.find(
    {'price' : { '$gt' : avg_price} },
    {'_id' : 0, 'name': 1, 'price':1})
for product in pricey:
    print(product)
```

```python
categorys = [
    {'$group': {'_id': '$category', 'amount_product': { '$sum': 1}}},
    {'$sort': {'amount_product': -1}}]

print("\nJumlah Produk per kategori")
category_result = products.aggregate(categorys)

for category in category_result:
    
    print(f"Kategori: {category['_id']}, Jumlah: {category['amount_product']}")

```

```python
top_5 = [{'$sort':{'salary':-1}},
         {'$group':{'_id': '$department','karyawan_jago':{'$push':{'name':'$name','salary':'$salary'}}}},
         {'$project':{'_id': 1, 'top_5_gaji':{'$slice':['$karyawan_jago',5]}}}]
hasil_top_5 = collection.aggregate(top_5)

for dept in hasil_top_5:
    print(f"Department {dept['_id']}")
    for karyawan in dept['top_5_gaji']:
        print(f"Nama: {karyawan['name']}, Gaji: {karyawan['salary']}")
```

```python
hasil = collection.delete_many({'age': {'$lt': 25}})

print(f"Jumlah karyawan yang dihapus: {hasil.deleted_count}")

new_employee = collection.find({}, {'_id': 0, 'name': 1, 'age': 1}).sort('age', 1)
for employee in new_employee:
    print(employee)
```

```python
report = [
    {'$group' : {'_id': '$department', 'total_gaji' : {'$sum': '$salary'}, 'rata_umur': {'$avg': '$age'}}},
    {'$project': {'_id': 0, 'department': '$_id', 'total_gaji':1, 'rata_umur':{'$round': ['$rata_umur', 2]}}},
    {'$sort':{'department': 1}}]

report_result = collection.aggregate(report)

for dept in report_result:
    print(f"Departemen: {dept['department']}")
    print(f"  Total Gaji: {dept['total_gaji']}")
    print(f"  Rata-rata Umur: {dept['rata_umur']}\n")
                
```
