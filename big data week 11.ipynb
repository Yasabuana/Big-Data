```python
# 1
from pymongo import MongoClient, InsertOne, UpdateOne, DeleteOne

client = MongoClient('mongodb://localhost:27017/')
db = client['university_db']
courses_collection = db['courses']

courses_collection.delete_many({})

operations =[
    InsertOne({'course': 'Math 101', 'enrollments': 30, 'department': 'Mathematics'}),
    InsertOne({'course': 'CS 102', 'enrollments': 25, 'department': 'Computer Science'}),
    InsertOne({'course': 'History 201', 'enrollments': 20, 'department': 'History'}),
    InsertOne({'course': 'Physics 202', 'enrollments': 15, 'department': 'Physics'}),
    InsertOne({'course': 'MS 203', 'enrollments': 10, 'department': 'Mathematics Science'}),
    UpdateOne({'course': 'MS 203'}, {'$set': {'enrollments': 27}})
]
courses_collection.bulk_write(operations)
print('Courses inserted successfully')

for course in courses_collection.find({}, {'_id':0}):
    print(course)
```

```python
# 2
for course in courses_collection.find({'enrollments': {'$gt': 20}}, {'_id': 0}):
    print(course)
```

```python
for course in courses_collection.find({'department': {'$in': ['Computer Science', 'Mathematics']}}, {'_id':0}):
    print(course)
# Jika ingin menampilkan semua kata yang mengandung kata yg sama ( contoh fisika sama teknik fisika )
```

```python
# 3
pipeline = [
    {'$group': {'_id': '$department', 'average_enroll' : {'$avg': '$enrollments'}}}]
for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
pipeline = [
    {'$group': {'_id': '$department', 'max_enroll' : {'$max': '$enrollments'}}}]

for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
# 4
pipeline = [
    {'$project': {'_id': 0, 'course_name' : '$course', 'department_name' : '$department', 'enrollments': 1}}]

for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
pipeline = [
    {'$addFields': {'_id' : 0, 'enrollment_category': {'$cond': [ {'$gt': ['$enrollments', 20]}, 'high','low']}}}]

for result in courses_collection.aggregate(pipeline):
    print(result)
```

```python
import pandas as pd


join = [
    {'$lookup' : { 'from' : 'students', 'localField' : 'course', 'foreignField' : 'enrolled_id', 'as' : 'many students'}}]

no_id = [
    {'$project' : {'_id': 0, 'many students._id' : 0}}]

merger = join + no_id 

result = list(courses_collection.aggregate(merger))

df = pd.DataFrame(result)
df['student_count'] = df['many students'].apply(lambda x: len(x))
df_new = df.drop(columns=['many students'])

print(df_new)
```
