```python
import pandas as pd

data = {
    'Name' : ['Alice', 'Bob', 'Charlie', 'David', None],
    'Age' : [24, 30, None, 22, 35],
    'Salary' : [48000, None, 57000, None, 60000]
}
df = pd.DataFrame(data)

example = df.fillna('Big Data')
print(example)

print('\nBefore Cleaning :\n ', df)

df['Age'].fillna(df['Age'].mean(), inplace=True)
print('Age :\n ',df['Age'])

df['Salary'].fillna(df['Salary'].median(), inplace=True)
print('Salary :\n', df['Salary'])
df.dropna(subset=['Name'], inplace=True)
print('\nAfter Cleaning :\n', df)
```

```python
import pandas as pd

data = pd.read_csv('Titanic-Dataset.csv')

print("Data Awal:")
print(data.head())

print("\nJumlah Missing Values Tiap Kolom:")
print(data.isnull().sum())
```

```python
from sklearn.preprocessing import MinMaxScaler

numeric_column = data.select_dtypes(include=['number']).columns
data[numeric_column] = data[numeric_column].fillna(data[numeric_column].median())
print(data.isnull().sum())

scaler = MinMaxScaler()
data[numeric_column] = scaler.fit_transform(data[numeric_column])

print("\nData setelah dinormalisasi")
print(data[numeric_column].head())
```

```python
from sklearn.preprocessing import LabelEncoder

category = data.select_dtypes(include=['object']).columns
data[category] = data[category].fillna('Unknown')
print(data.isnull().sum())

encoder = LabelEncoder()
for i in category:
    data[i] = encoder.fit_transform(data[i])

print("\nData setelah encoding")
print(data[category].head())
```

```python
before = len(data)
data_cleaned = data.drop_duplicates()
after = len(data_cleaned)
print(before)
print(after)
print(f"\nJumlah Duplikat Dihapus: {before - after}")
```
